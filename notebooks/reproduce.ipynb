# Reproduce the tables and figures

Run all cells to check the method on small exact examples and reproduce the displays from stored aggregate results. No simulation study is run. The notebook runs locally or in Colab. In Colab, it obtains the public repository automatically.


```python
from pathlib import Path
import sys, subprocess, time

REPOSITORY_URL = "https://github.com/calinadriancomes/rmst-linkage-variance-bounds.git"
REPOSITORY_REF = "main"

root = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "code/check.py").is_file()), None)
if root is None:
    if not REPOSITORY_URL:
        raise RuntimeError("The public repository address has not been set. Open this notebook in a local checkout.")
    root = Path.cwd() / "rmst-linkage-variance-bounds"
    if not root.exists():
        subprocess.run(["git", "clone", "--depth", "1", "--branch", REPOSITORY_REF,
                        REPOSITORY_URL, str(root)], check=True)
    if not (root / "code/check.py").is_file():
        raise RuntimeError("The repository does not contain code/check.py.")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r",
                    str(root / "requirements.txt")], check=True)
print("Repository:", root.name)
```

## Quick check

Verify the stable scientific checksums and small exact examples. No simulation draws are made.


```python
start = time.perf_counter()
subprocess.run([sys.executable, str(root / 'code/check.py')], check=True, cwd=root)
print(f'Quick check: {time.perf_counter() - start:.1f} seconds')
```

## Main tables and figures

Reproduce the two main tables and all ten figures from stored aggregate results. Exports are written to `outputs/`. The numerical inputs remain unchanged.


```python
start = time.perf_counter()
subprocess.run([sys.executable, str(root / 'code/reproduce.py'), '--output', str(root / 'outputs')], check=True, cwd=root)
print(f'Display reproduction: {time.perf_counter() - start:.1f} seconds')
```

## Read the outputs

The tables and four main figures are shown below. The six supplementary figures are also in `outputs/`. Coverage and width comparisons use their stated denominators; see the data guide for definitions.


```python
if 'IPython' in sys.modules:
    from IPython.display import display, Markdown, Image
    for table in sorted((root / 'outputs/tables').glob('*.md')):
        display(Markdown(table.read_text(encoding='utf-8')))
    for figure in sorted((root / 'outputs/main').glob('*.png')):
        display(Image(filename=str(figure)))
else:
    print('Tables and figures are in outputs/.')
```

## Save the displays

In Colab, run the final cell to download the output archive. Local users can open `outputs/` directly.


```python
import shutil
archive = shutil.make_archive(str(root / 'display_outputs'), 'zip', root / 'outputs')
if 'google.colab' in sys.modules:
    from google.colab import files
    files.download(archive)
else:
    print('Output archive:', Path(archive).name)
```
